```python
import nltk
import pandas as pd
```

```python
def elaborate_df(df):
    df_plus = df
    df_plus['q_tokens'] = df_plus['question'].apply(lambda x: nltk.tokenize.word_tokenize(x))
    df_plus['c_tokens'] = df_plus['context'].apply(lambda x: nltk.tokenize.word_tokenize(x))
    df_plus['q_tokens_len'] = df_plus['q_tokens'].apply(lambda x: len(x))
    df_plus['c_tokens_len'] = df_plus['c_tokens'].apply(lambda x: len(x))
    return df_plus
```

```python
elaborated_train = elaborate_df(df_train)
elaborated_val   = elaborate_df(df_val)
```

```python
def quantile_metrics(s):
    return [s.quantile(0.0).item(), s.quantile(0.25).item(), s.median().item(), s.quantile(0.75).item(), s.quantile(1.0).item()]
```

```python
for lang in df_train['lang'].unique():
    print(lang)
    print(f"\tTraining dataset quantiles:\t {quantile_metrics(elaborated_train[(elaborated_train['lang'] == lang)]['q_tokens_len'])}")
    print(f"\tValidation dataset quantiles:\t {quantile_metrics(elaborated_val[(elaborated_val['lang'] == lang)]['q_tokens_len'])}")
```
